```r
source("utilities.R")
# shffled SSMS, each dataset was shuffled 100 times
Austriaasref_Italy1_file_paths <- list.files(path = "data/imputed_ec/shuffle_network/Austriaasref_Italy1", pattern = "\\.csv$", full.names = TRUE)
Austriaasref_Italy2_file_paths <- list.files(path = "data/imputed_ec/shuffle_network/Austriaasref_Italy2", pattern = "\\.csv$", full.names = TRUE)
Italy1asref_Italy2_file_paths <- list.files(path = "data/imputed_ec/shuffle_network/Italy1asref_Italy2", pattern = "\\.csv$", full.names = TRUE)
Italy1asref_Austria_file_paths <- list.files(path = "data/imputed_ec/shuffle_network/Italy1asref_Austria", pattern = "\\.csv$", full.names = TRUE)
Italy2asref_Italy1_file_paths <- list.files(path = "data/imputed_ec/shuffle_network/Italy2asref_Italy1", pattern = "\\.csv$", full.names = TRUE)
Italy2asref_Austria_file_paths <- list.files(path = "data/imputed_ec/shuffle_network/Italy2asref_Austria", pattern = "\\.csv$", full.names = TRUE)

# 
Austriaasref_Italy1_list <- lapply(Austriaasref_Italy1_file_paths, function(path) {
  read.csv(path, row.names = 1, sep = "\t")
})
Austriaasref_Italy1_list <- lapply(Austriaasref_Italy1_list, function(df) df[order(rownames(df)), ])
#
Austriaasref_Italy2_list <- lapply(Austriaasref_Italy2_file_paths, function(path) {
  read.csv(path, row.names = 1, sep = "\t")
})
Austriaasref_Italy2_list <- lapply(Austriaasref_Italy2_list, function(df) df[order(rownames(df)), ])
#
Italy1asref_Italy2_list <- lapply(Italy1asref_Italy2_file_paths,function(path) {
  read.csv(path, row.names = 1, sep = "\t")
})
Italy1asref_Italy2_list <- lapply(Italy1asref_Italy2_list, function(df) df[order(rownames(df)), ])
#
Italy1asref_Austria_list <- lapply(Italy1asref_Austria_file_paths, function(path) {
  read.csv(path, row.names = 1, sep = "\t")
})
Italy1asref_Austria_list <- lapply(Italy1asref_Austria_list, function(df) df[order(rownames(df)), ])
#
Italy2asref_Italy1_list <- lapply(Italy2asref_Italy1_file_paths, function(path) {
  read.csv(path, row.names = 1, sep = "\t")
})
Italy2asref_Italy1_list <- lapply(Italy2asref_Italy1_list, function(df) df[order(rownames(df)), ])
#
Italy2asref_Austria_list <- lapply(Italy2asref_Austria_file_paths, function(path) {
  read.csv(path, row.names = 1, sep = "\t")
})
Italy2asref_Austria_list <- lapply(Italy2asref_Austria_list, function(df) df[order(rownames(df)), ])

# MGS
Italy2_mgs_level4ec <- read.delim('data/level4ec_and_pathway/mgs_level4ec_relab_unstratified/Italy2_mgs_level4ec_relab_unstratified.tsv', row.names = 1)
Italy1_mgs_level4ec <- read.delim('data/level4ec_and_pathway/mgs_level4ec_relab_unstratified/Italy1_mgs_level4ec_relab_unstratified.tsv', row.names = 1)
Austria_mgs_level4ec <- read.delim('data/level4ec_and_pathway/mgs_level4ec_relab_unstratified/Austria_mgs_level4ec_relab_unstratified.tsv', row.names = 1)

Italy1_mgs_level4ec <- Italy1_mgs_level4ec[order(rownames(Italy1_mgs_level4ec)), ]
Italy2_mgs_level4ec <- Italy2_mgs_level4ec[order(rownames(Italy2_mgs_level4ec)), ]
Austria_mgs_level4ec <- Austria_mgs_level4ec[order(rownames(Austria_mgs_level4ec)), ]

# imputed SSMS
predicted_level4ec_spiec_Austriaasref_Italy2 <- read.csv('data/imputed_ec/spieceasi/predicted_level4ec_spiec_Austriaasref_Italy2.csv', row.names = 1, sep = '\t')
predicted_level4ec_spiec_Austriaasref_Italy1 <- read.csv('data/imputed_ec/spieceasi/predicted_level4ec_spiec_Austriaasref_Italy1.csv', row.names = 1, sep = '\t')
predicted_level4ec_spiec_Italy1asref_Italy2 <- read.csv('data/imputed_ec/spieceasi/predicted_level4ec_spiec_Italy1asref_Italy2.csv', row.names = 1, sep = '\t')
predicted_level4ec_spiec_Italy2asref_Italy1 <- read.csv('data/imputed_ec/spieceasi/predicted_level4ec_spiec_Italy2asref_Italy1.csv', row.names = 1, sep = '\t')
predicted_level4ec_spiec_Italy1asref_Austria <- read.csv('data/imputed_ec/spieceasi/predicted_level4ec_spiec_Italy1asref_Austria.csv', row.names = 1, sep = '\t')
predicted_level4ec_spiec_Italy2asref_Austria <- read.csv('data/imputed_ec/spieceasi/predicted_level4ec_spiec_Italy2asref_Austria.csv', row.names = 1, sep = '\t')

predicted_level4ec_spiec_Austriaasref_Italy2 <- predicted_level4ec_spiec_Austriaasref_Italy2[order(rownames(predicted_level4ec_spiec_Austriaasref_Italy2)), ]
predicted_level4ec_spiec_Austriaasref_Italy1 <- predicted_level4ec_spiec_Austriaasref_Italy1[order(rownames(predicted_level4ec_spiec_Austriaasref_Italy1)), ]
predicted_level4ec_spiec_Italy1asref_Italy2 <- predicted_level4ec_spiec_Italy1asref_Italy2[order(rownames(predicted_level4ec_spiec_Italy1asref_Italy2)), ]
predicted_level4ec_spiec_Italy2asref_Italy1 <- predicted_level4ec_spiec_Italy2asref_Italy1[order(rownames(predicted_level4ec_spiec_Italy2asref_Italy1)), ]
predicted_level4ec_spiec_Italy1asref_Austria <- predicted_level4ec_spiec_Italy1asref_Austria[order(rownames(predicted_level4ec_spiec_Italy1asref_Austria)), ]
predicted_level4ec_spiec_Italy2asref_Austria <- predicted_level4ec_spiec_Italy2asref_Austria[order(rownames(predicted_level4ec_spiec_Italy2asref_Austria)), ]
```

```r
# calculate similarity between mgs and imputed data, and similarity between mgs and shuffled imputed data

# jaccard
true_Austriaasref_Italy2_jc <- compute_paired_similarity(Italy2_mgs_level4ec, predicted_level4ec_spiec_Austriaasref_Italy2, "jaccard")
shuffle_Austriaasref_Italy2_jc <- compute_multiple_similarity(Italy2_mgs_level4ec, Austriaasref_Italy2_list, "jaccard")

true_Austriaasref_Italy1_jc <- compute_paired_similarity(Italy1_mgs_level4ec, predicted_level4ec_spiec_Austriaasref_Italy1, "jaccard")
shuffle_Austriaasref_Italy1_jc <- compute_multiple_similarity(Italy1_mgs_level4ec, Austriaasref_Italy1_list, "jaccard")

true_Italy1asref_Italy2_jc <- compute_paired_similarity(Italy2_mgs_level4ec, predicted_level4ec_spiec_Italy1asref_Italy2, "jaccard")
shuffle_Italy1asref_Italy2_jc <- compute_multiple_similarity(Italy2_mgs_level4ec, Italy1asref_Italy2_list, "jaccard")

true_Italy1asref_Austria_jc <- compute_paired_similarity(Austria_mgs_level4ec, predicted_level4ec_spiec_Italy1asref_Austria, "jaccard")
shuffle_Italy1asref_Austria_jc  <- compute_multiple_similarity(Austria_mgs_level4ec, Italy1asref_Austria_list, "jaccard")

true_Italy2asref_Italy1_jc <- compute_paired_similarity(Italy1_mgs_level4ec, predicted_level4ec_spiec_Italy2asref_Italy1, "jaccard")
shuffle_Italy2asref_Italy1_jc <- compute_multiple_similarity(Italy1_mgs_level4ec, Italy2asref_Italy1_list, "jaccard")

true_Italy2asref_Austria_jc <- compute_paired_similarity(Austria_mgs_level4ec, predicted_level4ec_spiec_Italy2asref_Austria, "jaccard")
shuffle_Italy2asref_Austria_jc <- compute_multiple_similarity(Austria_mgs_level4ec, Italy2asref_Austria_list , "jaccard")


# spearman
true_Austriaasref_Italy2_sm <- compute_paired_similarity(Italy2_mgs_level4ec, predicted_level4ec_spiec_Austriaasref_Italy2, "spearman")
shuffle_Austriaasref_Italy2_sm <- compute_multiple_similarity(Italy2_mgs_level4ec, Austriaasref_Italy2_list, "spearman")

true_Austriaasref_Italy1_sm <- compute_paired_similarity(Italy1_mgs_level4ec, predicted_level4ec_spiec_Austriaasref_Italy1, "spearman")
shuffle_Austriaasref_Italy1_sm <- compute_multiple_similarity(Italy1_mgs_level4ec, Austriaasref_Italy1_list, "spearman")

true_Italy1asref_Italy2_sm <- compute_paired_similarity(Italy2_mgs_level4ec, predicted_level4ec_spiec_Italy1asref_Italy2, "spearman")
shuffle_Italy1asref_Italy2_sm <- compute_multiple_similarity(Italy2_mgs_level4ec, Italy1asref_Italy2_list, "spearman")

true_Italy1asref_Austria_sm <- compute_paired_similarity(Austria_mgs_level4ec, predicted_level4ec_spiec_Italy1asref_Austria, "spearman")
shuffle_Italy1asref_Austria_sm  <- compute_multiple_similarity(Austria_mgs_level4ec, Italy1asref_Austria_list, "spearman")

true_Italy2asref_Italy1_sm <- compute_paired_similarity(Italy1_mgs_level4ec, predicted_level4ec_spiec_Italy2asref_Italy1, "spearman")
shuffle_Italy2asref_Italy1_sm <- compute_multiple_similarity(Italy1_mgs_level4ec, Italy2asref_Italy1_list, "spearman")

true_Italy2asref_Austria_sm <- compute_paired_similarity(Austria_mgs_level4ec, predicted_level4ec_spiec_Italy2asref_Austria, "spearman")
shuffle_Italy2asref_Austria_sm <- compute_multiple_similarity(Austria_mgs_level4ec, Italy2asref_Austria_list , "spearman")
```

```r
# t test - jaccard
p.Austriaasref_Italy2_jc <- t.test(shuffle_Austriaasref_Italy2_jc,  mu = true_Austriaasref_Italy2_jc, alternative = "two.sided")$p.value
p.Austriaasref_Italy1_jc <- t.test(shuffle_Austriaasref_Italy1_jc,  mu = true_Austriaasref_Italy1_jc, alternative = "two.sided")$p.value
p.Italy1asref_Italy2_jc <- t.test(shuffle_Italy1asref_Italy2_jc,  mu = true_Italy1asref_Italy2_jc, alternative = "two.sided")$p.value
p.Italy1asref_Austria_jc <- t.test(shuffle_Italy1asref_Austria_jc,  mu = true_Italy1asref_Austria_jc, alternative = "two.sided")$p.value
p.Italy2asref_Italy1_jc <- t.test(shuffle_Italy2asref_Italy1_jc,  mu = true_Italy2asref_Italy1_jc, alternative = "two.sided")$p.value
p.Italy2asref_Austria_jc <- t.test(shuffle_Italy2asref_Austria_jc,  mu = true_Italy2asref_Austria_jc, alternative = "two.sided")$p.value

# t test - spearman
p.Austriaasref_Italy2_sm <- t.test(shuffle_Austriaasref_Italy2_sm,  mu = true_Austriaasref_Italy2_sm, alternative = "two.sided")$p.value
p.Austriaasref_Italy1_sm <- t.test(shuffle_Austriaasref_Italy1_sm,  mu = true_Austriaasref_Italy1_sm, alternative = "two.sided")$p.value
p.Italy1asref_Italy2_sm <- t.test(shuffle_Italy1asref_Italy2_sm,  mu = true_Italy1asref_Italy2_sm, alternative = "two.sided")$p.value
p.Italy1asref_Austria_sm <- t.test(shuffle_Italy1asref_Austria_sm,  mu = true_Italy1asref_Austria_sm, alternative = "two.sided")$p.value
p.Italy2asref_Italy1_sm <- t.test(shuffle_Italy2asref_Italy1_sm,  mu = true_Italy2asref_Italy1_sm, alternative = "two.sided")$p.value
p.Italy2asref_Austria_sm <- t.test(shuffle_Italy2asref_Austria_sm,  mu = true_Italy2asref_Austria_sm, alternative = "two.sided")$p.value
```

```r
# plot
library(ggplot2)

jc_plot <- data.frame(
  group = c("Austriaasref_Italy2", "Austriaasref_Italy1", "Italy1asref_Italy2","Italy1asref_Austria","Italy2asref_Italy1", "Italy2asref_Austria"),
  pvalue = c(p.Austriaasref_Italy2_jc, p.Austriaasref_Italy1_jc, p.Italy1asref_Italy2_jc, p.Italy1asref_Austria_jc, p.Italy2asref_Italy1_jc, p.Italy2asref_Austria_jc)
)

sm_plot <- data.frame(
  group = c("Austriaasref_Italy2", "Austriaasref_Italy1", "Italy1asref_Italy2","Italy1asref_Austria","Italy2asref_Italy1", "Italy2asref_Austria"),
  pvalue = c(p.Austriaasref_Italy2_sm, p.Austriaasref_Italy1_sm, p.Italy1asref_Italy2_sm, p.Italy1asref_Austria_sm, p.Italy2asref_Italy1_sm, p.Italy2asref_Austria_sm)
)

str(sm_plot)

# add column：-log10(pvalue)
jc_plot$log10_pvalue <- -log10(jc_plot$pvalue)
sm_plot$log10_pvalue <- -log10(sm_plot$pvalue)
# plot
ggplot(jc_plot, aes(x = group, y = log10_pvalue)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_hline(yintercept = -log10(0.05), linetype = "dashed", color = "red") +
  labs(y = expression(-log[10](p~value)), x = "Group") +
  theme_minimal() +
  theme(panel.grid = element_blank()) 

ggplot(sm_plot, aes(x = group, y = log10_pvalue)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_hline(yintercept = -log10(0.05), linetype = "dashed", color = "red") +
  labs(y = expression(-log[10](p~value)), x = "Group") +
  theme_minimal() +
  theme(panel.grid = element_blank()) 
```
